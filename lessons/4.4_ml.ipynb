## Part 1: Introduction to Markov Chains

- **Definition: Markov Property**

    - Motivation: The goal is to model an underlying system, mathematically capturing its behavior as closely as possible. Specifically, a Markov Chain is used to model complex memoryless systems.

    - Given the present state of the system, past states have no influence on the future state.

    - Mathematically, we can denote this as:
    $$\mathbb{P}[X_{n+1} = x_{n+1} \mid X_{n} = x_{n}] = \mathbb{P}[X_{n+1} = x_{n+1} \mid X_{0} = x_0,\, X_{1} = x_1,\, \ldots,\, X_{n} = x_{n}] \quad \forall\, n \text{ and } \forall\, x$$


- **The One-Step Transition Probability Matrix $P$**

    - Let $\mathcal{S}$ denote the state space. For example, $\mathcal{S} = \{0, 1, 2, \ldots, n\}$, which represents a Markov Chain having $n + 1$ states. It is conventional to index from zero.

    - Let $|\mathcal{S}| < \infty$ (just the mathematical way of saying the Markov Chain has a finite state space). Then, for a Markov Chain having $d + 1$ states we have: $$P = \begin{pmatrix} P(0,0) & P(0,1) & \cdots & P(0,d) \\ P(1,0) & P(1,1) & \cdots & P(1,d) \\ \vdots & \vdots & \ddots & \vdots \\ P(d,0) & P(d,1) & \cdots & P(d,d) \end{pmatrix}$$

    - Notice that $P$ contains the one-step transition probabilities between every pair of states.

    - Question: What if we'd like to know the probabilities for two steps?

        - Let $x,y,z \in \mathcal{S}$. Then: $$P^2(x,y) = \sum_{z} P(x,z) \cdot P(z,y)$$

        - Note, the superscript $2$ in $P^2(x,y)$ represents the two-step transition probability.

        - We are applying the Law of Total Probability.

        - Think of the _multiplication_ of the two transition probabilities (each term in the summation) as an _and_ statement: the probability of going to some intermediate state $z$ given you started at $x$, _and_ then from $z$ the probability of reaching $y$.

        - Think of the _summation_ as an _or_ statement: we can travel from $x$ to $z_0$ then to $y$, _or_ from $x$ to $z_1$ then to $y$, and so on for all $z \in \mathcal{S}$.

        - Notice this formula mirrors the dot product, which means there is a direct relationship to matrix multiplication.

    - Question: What if we'd like to generalize the probabilities for $n$ steps?

        - Let $x,y,z \in \mathcal{S}$. Then: $$P^{n}(x,y) = \sum_{z} P^{n-1}(x,z) \cdot P(z,y)$$

        - Note, the superscript $n$ in $P^{n}(x,y)$ represents the $n$-step transition probability. Intuitively, this means the probability of being in state $y$ after $n$ steps given you started in state $x$ is $P^{n}(x,y)$.

        - However, this formula only represents a transition probability from state-to-state. What if we wanted the $n$-step transition probabilities for all states? This is simply $P$ raised to the $n$-th power: $$P^{n} = \underbrace{P \cdot P \cdots P}_{n \text{ times}}$$

        - From $P^{n}$ you can extract the specific probabilities (matrix entries) you are interested in.

    - Properties of $P$

        - The rows sum to one: $\sum_{j=0}^{d} P_{i,j} = 1 \quad \forall\, i$

        - $P$ is a square matrix: $(d+1) \times (d+1)$

        - $P$ is stochastic, meaning each entry is a valid probability: $0 \le P_{i,j} \le 1$

        - $P$ is guaranteed to have an eigenvalue of $1$

- **The Initial Distribution $\mathbf{\pi}$**

    - Let $\mathbf{\pi}_0$ be the initial distribution for a Markov Chain with $d+1$ states: $$\mathbf{\pi}_0 = (\pi_0(0),\, \pi_0(1),\, \ldots,\, \pi_0(d))$$

    - The initial distribution, at time zero, can be thought of as the probability of the state the system starts in.

    - The distribution at time $n$ can be thought of as the probability of the state the system is in after $n$ steps: $$\mathbf{\pi}_n = \mathbf{\pi}_0 \cdot P^n$$

- **Definition: Steady State**

    - Motivating Question: What is the long-term probability of being at each state?

    - Suppose that a stationary distribution $\mathbf{\pi}$ exists. Then, $$\lim_{n \to \infty} [P^n(x,y)] = \pi(y), \quad x,y \in \mathcal{S}$$

    - Notice, the limit converges to $\pi(y)$, and so we see that steady-state for state $y$ is independent of the state you started in.

- **Definition: Stationary Distribution**

    - A distribution is stationary if the probability distribution over states remains the same after a transition, expressed as $\mathbf{\pi} = \pi P $, where $P$ is the transition matrix.

- **Time-Homogeneous**

    - As you may have observed, the transition probabilities in our $P$ matrix have not changed over time.

    - A Markov Chain is said to be time-homogeneous when the transition probabilities remain constant as time progresses. This property can be denoted as:
    $$\mathbb{P}[X_{n+1} = i \mid X_{n} = j] = \mathbb{P}[X_{1} = i \mid X_{0} = j] \quad \forall\, n$$

    - An intuitive way to think about this: without time-homogeneity, moving from one state to another would depend not only on the current state (the Markov Property), but also on the current time-step.

    - This property is crucial for the analysis of Steady States and Stationary Distributions.

    - For some games, examples, or real-life applications, time-homogeneity does not hold.

- **Time-Inhomogeneous**

    - Although sometimes more realistic, it is not common in literature or textbooks because it is difficult to derive interesting results if a system is assumed to be time-inhomogeneous.

    - Transition Probabilities

        - Let $x,t$ represent some state $x$ at time $t$, and $y,s$ represent some state $y$ at time $s$. Then our old transition function $P(x,y)$ can be rewritten: $$K_t^s(x, y) = \mathbb{P}[X_s = y \mid X_t = x]$$

        - In this setting, we let $t=0$, and $P$ becomes: $$K_0^s = \begin{pmatrix} K_0^s(0,0) & K_0^s(0,1) & \cdots & K_0^s(0,d) \\ K_0^s(1,0) & K_0^s(1,1) & \cdots & K_0^s(1,d) \\ \vdots & \vdots & \ddots & \vdots \\ K_0^s(d,0) & K_0^s(d,1) & \cdots & K_0^s(d,d) \end{pmatrix}$$

    - Generalizing to $n$-step transition probabilities also becomes more convoluted: $$K_t^s = K_t^{t+1} \cdot K_{t+1}^{t+2} \cdot K_{t+2}^{t+3} \cdots K_{s-1}^{s}$$

    - Notice, unlike the time-homogeneous case where we could simplify to $P^n$, here we cannot collapse the product into a single matrix power because each $K_t^{t+1}$ may be different.

```python
import numpy as np
import pandas as pd
from datetime import datetime, timedelta
```

```python
# Time-Inhomogenous Example

# State space
states = {'snow', 'rain', 'overcast', 'sun'}

# Matrix format
"""             snow  rain  overcast  sun
    snow        ...
    rain
    overcast
    sun
"""

# Fall weather matrix
F = np.array([
    [0.30, 0.20, 0.40, 0.10],
    [0.20, 0.30, 0.40, 0.10],
    [0.10, 0.25, 0.50, 0.15],
    [0.05, 0.15, 0.60, 0.20]
])

# Winter weather matrix
W = np.array([
    [0.55, 0.20, 0.20, 0.05],
    [0.30, 0.40, 0.25, 0.05],
    [0.20, 0.30, 0.40, 0.10],
    [0.10, 0.20, 0.50, 0.20]
])

# Spring weather matrix
S = np.array([
    [0.10, 0.30, 0.40, 0.20],
    [0.05, 0.35, 0.40, 0.20],
    [0.02, 0.30, 0.38, 0.30],
    [0.01, 0.24, 0.35, 0.40]
])

# Summer weather matrix
M = np.array([
    [0.05, 0.10, 0.35, 0.50],
    [0.00, 0.20, 0.35, 0.45],
    [0.00, 0.15, 0.30, 0.55],
    [0.00, 0.10, 0.20, 0.70]
])

spring_start = pd.Timestamp('2025-03-20')
summer_start = pd.Timestamp('2025-06-21')
fall_start   = pd.Timestamp('2025-09-22')
winter_start = pd.Timestamp('2025-12-21')

today = pd.Timestamp(datetime.today())
```

```python
# Initial distribution

pi_0 = np.array([0, 0, 1, 0]) # Start in 'overcast'
```

```python
# Matrix multiplication function

def K_sequence(days=100):

    end_date = today + timedelta(days=days)

    K = np.eye(4)

    for day in pd.date_range(today, end_date, inclusive='right'):

        # Spring
        if (spring_start <= day) and (day < summer_start):
            K @= S
        
        # Summer
        elif (summer_start <= day) and (day < fall_start):
            K @= M

        # Fall
        elif (fall_start <= day) and (day < winter_start):
            K @= F

        # Winter
        else:
            K @= W

    return K
```

```python
# What is the probability of 'snow' 100 days from now?

K = K_sequence(100)

pi_n = pi_0 @ K

print(f"{pi_n[0]*100:.2f}%")
```

```python
# What is the probability of 'sun' 100 days from now?

print(f"{pi_n[3]*100:.2f}%")
```

```python
# What is the probability of 'snow' 3 days from now?

K = K_sequence(3)

pi_n = pi_0 @ K

print(f"{pi_n[0]*100:.2f}%")
```

```python
# What is the probability of 'sun' 3 days from now?

print(f"{pi_n[3]*100:.2f}%")
```

## Part 2: Introduction to Hidden Markov Models


- **Definition: Hidden Markov Model**

    - In the context of Machine Learning, a Hidden Markov Model is an _unsupervised_ generative model that assumes the observed data is produced by an underlying sequence of hidden states.

    - Unsupervised means data labels are not provided; instead, the algorithm discovers the underlying structure or patterns in the data on its own.

    - A Hidden Markov Model (HMM) is a probabilistic model representing probability distributions over _sequences of observations_.

    - For notation, we can define an HMM as: $$\text{HMM} = \lambda = (N, T, P, E, \vec{\pi})$$

    - Where:
        - $N$ is the number of states, and $P$ is the (now familiar) accompanying
        transition probability matrix. Note, $P$ is $N \times N$.

        - $T$ is the number of observations, and $E$ is the (new) accompanying
        emission probability matrix. Note, $E$ is $N \times T$.

        - Fortunately, usually we will have $N = T$, meaning there is exactly one
        observation for each hidden state.

        - Lastly, we have the (now familiar) initial distribution vector,
        $\boldsymbol{\pi}$, which contains $N$ entries.


- **Transition Probabilities**

$$\mathbb{P}[X_{t+1} = s_i \mid X_{t} = s_j] = P_{i,j} = p(s_{t+1} \mid s_{t})$$


- **Emission Probabilities**

    - The probability that a specific observation will occur given that the system is currently in a specific hidden state.

    - These are the link between hidden states and observations.

$$\mathbb{P}[O_{t} = o_i \mid X_{t} = s_j] = E_{i,j} = e(o_{t} \mid s_{t})$$


- **Marginal Probability of a Sequence of States**

    - We are given a _specific_ sequence of states $\bar{s} = \{s_t\}_{t=1}^T$; how would we find the probability of this sequence occurring?

    - Also, we have a model $\lambda = (N, T, P, E, \vec{\pi})$.

    - For a $\text{HMM}$, the Markov property holds, meaning each state depends only on the previous state, and so: $$\mathbb{P}[\bar{s} \mid \lambda] = \vec{\pi}(s_1) \cdot \prod_{t=1}^T p(s_{t+1} \mid s_{t}) $$


- **Conditional Probability of a Sequence of Observations Given a Sequence of States**

    - We are given a _specific_ sequence of states $\bar{s} = \{s_t\}_{t=1}^T$ and a _specific_ sequence of observations $\bar{o} = \{o_t\}_{t=1}^T$; how would we find the probability of the sequence of observations occurring given the sequence of states?

    - Also, we have a model $\lambda = (N, T, P, E, \vec{\pi})$.

    - It is assumed that the observation $o_t$ at time $t$ is independent of all states at other times, and so: $$\mathbb{P}[\bar{o} \mid \bar{s}, \lambda] = \prod_{t=1}^T e(o_{t} \mid s_{t})$$


- **Marginal Probability of a Sequence of States _and_ a Sequence of Observations**

    - We are given a _specific_ sequence of states $\bar{s} = \{s_t\}_{t=1}^T$ and a _specific_ sequence of observations $\bar{o} = \{o_t\}_{t=1}^T$; how would we find the probability of $\bar{s}$ _and_ $\bar{o}$ occurring?

    - Also, we have a model $\lambda = (N, T, P, E, \vec{\pi})$.
    
    - Notice: $$\mathbb{P}[\bar{s}, \bar{o} \mid \lambda] = \mathbb{P}[\bar{o} \mid \bar{s}, \lambda] \cdot \mathbb{P}[\bar{s} \mid \lambda]$$

    - For a $\text{HMM}$, the Markov property holds, meaning each state depends only on the previous state. Additionally, it is assumed that the observation $o_t$ at time $t$ is independent of all states at other times:
    $$\mathbb{P}[\bar{s}, \bar{o} \mid \lambda] = \vec{\pi}(s_1) \cdot e(o_1 \mid s_1) \cdot \prod_{t=2}^{T} p(s_{t} \mid s_{t-1}) \cdot e(o_{t} \mid s_{t}) \tag{1}$$


- **Marginal Probability of a Sequence of Observations Under a Particular Model**

    - We seek the probability of a specific sequence of observations under a particular model.

    - To accomplish this we can use $(1)$ and sum over all possible state sequences: $$\mathbb{P}[\bar{o} \mid \lambda] = \sum_{\bar{s}} \vec{\pi}(s_1) \cdot e(o_1 \mid s_1) \cdot \prod_{t=2}^{T} p(s_{t} \mid s_{t-1}) \cdot e(o_{t} \mid s_{t})$$

    - Notice that each observation $o_t$ is fixed, but each state $s_t$ is varying.

    - Although this answers our question, it is computationally expensive; the number of possible state sequences grows exponentially in $T$. So, how can we do this efficiently?
